## 1. 감정 분석


```python
from transformers import pipeline
sentiment = pipeline("text-classification", model="daekeun-ml/koelectra-small-v3-nsmc")
text = "배송도 빠르고 상품도 정말 신선해서 만족합니다."
result = sentiment(text)
print("입력:", text)
print("결과:", result)
```

## 2. 기계 번역 — 한국어 → 영어


```python
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-ko-en")
text = "바다몰은 신선한 수산물을 판매하는 온라인 쇼핑몰입니다."
result = translator(text, max_length=100)
print("입력:", text)
print("번역:", result[0]["translation_text"])
```

## 3. 문서 요약


```python
summarizer = pipeline("summarization", model="EbanLee/kobart-summary-v2")
text = """온라인 수산물 쇼핑몰은 소비자가 시장에 직접 방문하지 않아도 다양한 수산물을 비교하고 주문할 수 있다는 장점이 있다. 특히 냉동과 냉장 배송 기술이 발달하면서 새우, 연어, 조개류와 같은 상품도 비교적 안정적으로 배송할 수 있게 되었다. 소비자는 상품의 원산지와 중량, 가격, 배송 조건을 확인한 뒤 주문하며, 판매자는 상품 정보를 정확하게 제공해야 한다. 또한 온라인 쇼핑몰은 검색, 장바구니, 주문 조회와 같은 기능을 제공하여 구매 과정을 편리하게 만든다."""
result = summarizer(text, max_length=90, min_length=20, do_sample=False)
print("원문:", text)
print("요약:", result[0]["summary_text"])
```
